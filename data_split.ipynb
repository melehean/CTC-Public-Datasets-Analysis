```python
import scanpy as sc
import pandas as pd
from sklearn.model_selection import train_test_split
```

```python
files = [
    "./ready_data/ctc_normalized_data_fixed.h5ad",
    "./ready_data/combined_tumor_data_fixed.h5ad",
    "./ready_data/combined_healthy_data_fixed.h5ad"
]
label_list = ["CTC", "Tumor", "PBMC"]

print("Skanowanie plików na dysku...")
adatas_backed = {}
for f, label in zip(files, label_list):
    a = sc.read_h5ad(f, backed='r')
    a.obs_names = [name.replace("-", "_") for name in a.obs_names]
    a.obs['batch'] = [ "_".join(name.split('_')[:2]) for name in a.obs_names ]
    a.strings_to_categoricals()
    adatas_backed[label] = a
```

```python
healthy_batch = adatas_backed["PBMC"].obs["batch"]
train_healthy_indices = healthy_batch[healthy_batch == "E_HCAD"].index
val_healthy_indices = healthy_batch[healthy_batch.isin(["pbmc_68k", "pbmc_3k", "pbmc_4k"])].index
test_healthy_indices = healthy_batch[healthy_batch.isin(["our_healthy", "pbmc_10k", "pbmc_8k",
                                                         "pbmc_6k"])].index
```

```python
adatas_backed["PBMC"][train_healthy_indices].copy(filename="./ready_data/train_healthy_fixed.h5ad")
adatas_backed["PBMC"][val_healthy_indices].copy(filename="./ready_data/val_healthy_fixed.h5ad")
adatas_backed["PBMC"][test_healthy_indices].copy(filename="./ready_data/test_healthy_fixed.h5ad")
```

```python
tumor_batch = adatas_backed["Tumor"].obs["batch"]
train_tumor_indices = tumor_batch[tumor_batch.isin(["gse_194247", "gse_125449", "gse_181294",
                                                          "our_spikes"])].index
val_tumor_indices = tumor_batch[tumor_batch == "gse_131907"].index
test_tumor_indices = tumor_batch[tumor_batch == "gse_176078"].index
```

```python
adatas_backed["Tumor"][train_tumor_indices].copy(filename="./ready_data/train_tumor_fixed.h5ad")
adatas_backed["Tumor"][val_tumor_indices].copy(filename="./ready_data/val_tumor_fixed.h5ad")
adatas_backed["Tumor"][test_tumor_indices].copy(filename="./ready_data/test_tumor_fixed.h5ad")
```
